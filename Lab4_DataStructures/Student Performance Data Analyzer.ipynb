```python
last_name = "BASAS"
favorite_artist = "BRUNO MARS"
seed_num = 3492

positions = [ord(c) - 64 for c in favorite_artist if c.isalpha()]

students = [f"{last_name}-S{i}" for i in range(1, 6)]

subjects = [
    "PROGRAMMING",
    "MATHEMATICS",
    "ELECTRONICS",
    "PHYSICS",
    "ENGLISH"
]

records = []

for i in range(5):
    score = 50 + ((seed_num + positions[i % len(positions)]) % 51)

    records.append({
        "Student": students[i],
        "Subject": subjects[i],
        "Score": score
    })

# Repeated record
records.append(records[0])

# Incomplete record
records.append({
    "Student": students[1],
    "Subject": "PROGRAMMING",
    "Score": None
})

print("=== GENERATED PERFORMANCE RECORDS ===")
for r in records:
    print(r)

def performance_category(score):
    if score is None:
        return "INCOMPLETE"
    elif score >= 90:
        return "EXCELLENT"
    elif score >= 80:
        return "VERY GOOD"
    elif score >= 70:
        return "GOOD"
    elif score >= 60:
        return "FAIR"
    else:
        return "POOR"

print("\n=== PERFORMANCE CATEGORIES ===")
for r in records:
    print(
        r["Student"],
        r["Subject"],
        performance_category(r["Score"])
    )

print("\n=== REPEATED RECORDS ===")
seen = set()

for r in records:

    if r["Score"] is None:
        continue

    item = (r["Student"], r["Subject"], r["Score"])

    if item in seen:
        print(item)
    else:
        seen.add(item)

print("\n=== INCOMPLETE RECORDS ===")
for r in records:
    if r["Score"] is None:
        print(r)

valid_scores = [
    r["Score"]
    for r in records
    if r["Score"] is not None
]

print("\n=== STUDENT SUMMARY ===")
print("Average Score:", round(sum(valid_scores) / len(valid_scores), 2))
print("Highest Score:", max(valid_scores))
print("Lowest Score:", min(valid_scores))

print("\n=== FINAL REPORT ===")
print("Total Records:", len(records))
print("Valid Scores:", len(valid_scores))
```
