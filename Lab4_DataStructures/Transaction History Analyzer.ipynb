```python
from collections import defaultdict

last_name = "BASAS"
favorite_artist = "BRUNO MARS"
seed_num = 3492

positions = [ord(c) - 64 for c in favorite_artist if c.isalpha()]

customers = [f"{last_name}-C{i}" for i in range(1, 6)]

products = {
    "NOTEBOOK": 50,
    "PEN": 15,
    "USB": 350,
    "HEADPHONES": 800,
    "CALCULATOR": 250
}

product_names = list(products.keys())

transactions = []

for i in range(5):

    quantity = (seed_num % 5) + positions[i % len(positions)]

    product = product_names[i]

    price = products[product]

    total = quantity * price

    transactions.append({
        "Customer": customers[i],
        "Product": product,
        "Quantity": quantity,
        "Price": price,
        "TransactionValue": total
    })

# Repeated purchase
transactions.append(transactions[0])

print("=== GENERATED TRANSACTION RECORDS ===")

for t in transactions:
    print(t)

overall_total = 0

print("\n=== TRANSACTION VALUES ===")

for t in transactions:
    overall_total += t["TransactionValue"]

    print(
        t["Customer"],
        t["Product"],
        "=",
        t["TransactionValue"]
    )

print("\n=== REPEATED TRANSACTIONS ===")

seen = set()

for t in transactions:

    item = (
        t["Customer"],
        t["Product"],
        t["Quantity"]
    )

    if item in seen:
        print(item)
    else:
        seen.add(item)

print("\n=== CUSTOMER SUMMARIES ===")

customer_totals = defaultdict(int)

for t in transactions:
    customer_totals[t["Customer"]] += t["TransactionValue"]

for customer, total in customer_totals.items():
    print(customer, total)

print("\n=== PRODUCT SUMMARIES ===")

product_totals = defaultdict(int)

for t in transactions:
    product_totals[t["Product"]] += t["Quantity"]

for product, quantity in product_totals.items():
    print(product, quantity)

print("\n=== OVERALL TRANSACTION SUMMARY ===")
print("Total Transaction Value:", overall_total)
print("Total Transactions:", len(transactions))
```
