```python
from collections import defaultdict

last_name = "BASAS"
favorite_artist = "BRUNO MARS"
seed_num = 3492

positions = [ord(c) - 64 for c in favorite_artist if c.isalpha()]

sources = [f"{last_name}-T{i}" for i in range(1, 6)]
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

records = []

for i in range(5):
    value = seed_num + positions[i % len(positions)]

    records.append({
        "Source": sources[i],
        "Category": categories[i],
        "Value": value
    })

# Additional measurements
records.append({
    "Source": sources[0],
    "Category": "TEMPERATURE",
    "Value": seed_num + 20
})

records.append({
    "Source": sources[1],
    "Category": "VOLTAGE",
    "Value": seed_num + 25
})

# Repeated record
records.append(records[0])

print("=== GENERATED TEST DATA ===")
for r in records:
    print(r)

organized = defaultdict(list)

for r in records:
    organized[(r["Source"], r["Category"])].append(r["Value"])

print("\n=== ORGANIZED TEST DATA ===")
for key, values in organized.items():
    print(key, values)

print("\n=== REPEATED RECORDS ===")
seen = set()

for r in records:
    item = (r["Source"], r["Category"], r["Value"])

    if item in seen:
        print(item)
    else:
        seen.add(item)

values = [r["Value"] for r in records]

print("\n=== NUMERICAL SUMMARY ===")
print("Count:", len(values))
print("Minimum:", min(values))
print("Maximum:", max(values))
print("Average:", round(sum(values) / len(values), 2))

print("\n=== FINAL SUMMARY ===")
print("Total Records:", len(records))
print("Distinct Sources:", len(sources))
print("Distinct Categories:", len(categories))
```
